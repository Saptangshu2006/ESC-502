**Algorithm**

**Checking for Prime Numbers:**

1. First, we check base case: If n <= 1, it is not prime. If n=2 or 3, it is prime.

2. We eliminate even numbers. If n > 2, it is not prime.

3. We test odd divisors upto (n)^0.5.

4. If no number divides n evenly, n is prime.

```python
import math
```

```python
def is_prime(n):
    if n <= 1:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0:
        return False
    limit = int(math.isqrt(n))
    for divisor in range(3, limit + 1, 2):
        if n % divisor == 0:
            return False
    return True
```

```python
test = [1, 2, 17, 25, 29, 100]
for num in test:
    print(f"{num}: {'Prime' if is_prime(num) else 'Not prime'}")
```
